Q2: Write a program that accepts a list from user. Your program should be reverse the content of list and display it. Do not use reverse() method.

```python
elements = [1,2,3,4,5,6,7,8,9]

reversed_list = []
for i in range(len(elements) - 1, -1, -1):
    reversed_list.append(elements[i])

print("Original list:", elements)
print("Reversed list:", reversed_list)
```

Q3: Find and display the largest number of a list without using built-in function max(). Your program should ask the user to input values in list from keyboard.

```python

elements = input("1,2,3,4,5,6,7,8,9: ").split()
numbers = []
for value in elements:
    numbers.append(int(value))

largest = numbers[0]   
for num in numbers:
    if num > largest:
        largest = num

print("List:", numbers)
print("Largest number is:", largest)
```

Q7:  Write a program with a function that accepts a string from keyboard and create a new string after converting character of each word capitalized. For instance, if the sentence is "stop and smell the roses." the output should be "stop And smell the Roses".

```python
def capitalize_words(sentence):
    words = sentence.split()
    
    new_words = []
    for word in words:
        capitalized_word = word.capitalize()
        new_words.append(capitalized_word)
    
    new_sentence = " ".join(new_words)
    return new_sentence


sentence = input("Enter a sentence: ")
result = capitalize_words(sentence)

print("Original sentence:", sentence)
print("Capitalized sentence:", result)
```

Q8: Find the sumof each row of matrix of size m x n. for Example for the following matrix output will be like this :

2   11  7   12
5   2   9   15
8   3   10  42

sum of row 1 = 32

sum of row 2 = 31

sum or row 3 = 63

```python
matrix = [
    [2, 11, 7, 12],
    [5, 2, 9, 15],
    [8, 3, 10, 42]
]

row_number = 1
for row in matrix:
    row_sum = 0
    for element in row:
        row_sum = row_sum + element
    print(f"sum of row {row_number} = {row_sum}")
    row_number = row_number + 1
```

Q9: Write a program to add two matrices of size n x m.

```python

matrix1 = [
    [2, 11, 7],
    [5, 2, 9],
    [8, 3, 10]
]

matrix2 = [
    [1, 4, 6],
    [3, 7, 2],
    [9, 5, 1]
]

n = len(matrix1)        
m = len(matrix1[0])      

result = []

for i in range(n):
    row = []
    for j in range(m):
        sum_value = matrix1[i][j] + matrix2[i][j]
        row.append(sum_value)
    result.append(row)

print("Matrix 1:")
for row in matrix1:
    print(row)

print("\nMatrix 2:")
for row in matrix2:
    print(row)

print("\nSum of Matrix 1 and Matrix 2:")
for row in result:
    print(row)

```

Q10: Write a program to multiply two matrices.

```python

matrix1 = [
    [1, 2, 3],
    [4, 5, 6]
]

matrix2 = [
    [7, 8],
    [9, 10],
    [11, 12]
]

r1 = len(matrix1)        
c1 = len(matrix1[0])     

r2 = len(matrix2)        
c2 = len(matrix2[0])     

if c1 != r2:
    print("Matrix multiplication not possible! (columns of matrix1 must equal rows of matrix2)")
else:
    
    result = []
    for i in range(r1):
        row = [0] * c2
        result.append(row)

    for i in range(r1):
        for j in range(c2):
            sum_value = 0
            for k in range(c1):
                sum_value = sum_value + matrix1[i][k] * matrix2[k][j]
            result[i][j] = sum_value

    print("Matrix 1:")
    for row in matrix1:
        print(row)

    print("\nMatrix 2:")
    for row in matrix2:
        print(row)

    print("\nMultiplication of Matrix 1 and Matrix 2:")
    for row in result:
        print(row)
```
