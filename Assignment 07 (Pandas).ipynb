8) Sort the DataFrame by Salary in descending order, and then by Age in ascending order. Also print the top 3 highest-paid employees using nlargest().

```python
df = pd.DataFrame({
    "Name": ["Ali", "Sara", "Ahmed", "Fatima", "Usman", "Ayesha", "Bilal", "Hina"],
    "Age": [28, 34, 25, 41, 30, 27, 36, 29],
    "City": ["Karachi", "Lahore", "Islamabad", "Karachi", "Peshawar", "Lahore", "Quetta", "Karachi"],
    "Salary": [65000, 85000, 55000, 120000, 70000, 85000, 95000, 70000]
})

# Sort by Salary (descending), then Age (ascending)
sorted_df = df.sort_values(by=["Salary", "Age"], ascending=[False, True])
print("Sorted by Salary (desc), then Age (asc):\n", sorted_df)

# Top 3 highest-paid employees
top3 = df.nlargest(3, "Salary")
print("\nTop 3 highest-paid:\n", top3)
```

9) Using groupby, find the average salary per City, the number of employees per City, and the maximum salary per City. Show the result using agg() with multiple functions.

```python
df = pd.DataFrame({
    "Name": ["Ali", "Sara", "Ahmed", "Fatima", "Usman", "Ayesha", "Bilal", "Hina"],
    "Age": [28, 34, 25, 41, 30, 27, 36, 29],
    "City": ["Karachi", "Lahore", "Islamabad", "Karachi", "Peshawar", "Lahore", "Karachi", "Lahore"],
    "Salary": [65000, 85000, 55000, 120000, 70000, 60000, 95000, 72000]
})

# Individual groupby results
print("Average salary per City:\n", df.groupby("City")["Salary"].mean())
print("\nEmployees per City:\n", df.groupby("City")["Name"].count())
print("\nMax salary per City:\n", df.groupby("City")["Salary"].max())

# All together using agg()
result = df.groupby("City")["Salary"].agg(["mean", "count", "max"])
print("\nCombined with agg():\n", result)

# Same thing with custom column names (named aggregation)
result2 = df.groupby("City").agg(
    Avg_Salary=("Salary", "mean"),
    Employees=("Name", "count"),
    Max_Salary=("Salary", "max")
)
print("\nNamed aggregation:\n", result2)
```

10) Create two DataFrames (employees and departments) with a common column 'DeptID'. Merge them using inner and left joins, then concatenate two DataFrames of the same structure using pd.concat() and remove duplicate rows.

```python
# Merge
employees = pd.DataFrame({
    "EmpID": [1, 2, 3, 4, 5],
    "Name": ["Ali", "Sara", "Ahmed", "Fatima", "Usman"],
    "DeptID": [10, 20, 30, 10, 40]
})

departments = pd.DataFrame({
    "DeptID": [10, 20, 30, 50],
    "DeptName": ["HR", "IT", "Finance", "Marketing"]
})

inner = pd.merge(employees, departments, on="DeptID", how="inner")
left = pd.merge(employees, departments, on="DeptID", how="left")

print("Inner join:\n", inner)
print("\nLeft join:\n", left)

# Concat + remove duplicate
df1 = pd.DataFrame({
    "Name": ["Ali", "Sara", "Ahmed"],
    "City": ["Karachi", "Lahore", "Islamabad"]
})
df2 = pd.DataFrame({
    "Name": ["Sara", "Usman", "Ali"],
    "City": ["Lahore", "Peshawar", "Karachi"]
})

combined = pd.concat([df1, df2], ignore_index=True)
print("\nAfter concat:\n", combined)

unique = combined.drop_duplicates()
print("\nAfter drop_duplicates:\n", unique)
```
