Q-4 Count Frequency of Numbers
    numbers = [1,2,2,3,3,3,4,4,4,4]
    create frequency dictionary.

```python

numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

frequency = {}

for num in numbers:
    if num in frequency:
    
        frequency[num] = frequency[num] + 1
    else:
       
        frequency[num] = 1

print("Numbers list:", numbers)
print("Frequency dictionary:", frequency)
```

Q-5 Word Counter

Take sentence input and count word frequency.

```python

sentence = input("Enter a sentence: ")

words = sentence.split()

frequency = {}

for word in words:
    if word in frequency:
        
        frequency[word] = frequency[word] + 1
    else:
        
        frequency[word] = 1

print("Sentence:", sentence)
print("Word frequency dictionary:", frequency)
```

Q-6 Students Marks and Average
    marks = {"Math" :80, "English":70, "Physics": 90}
Calculate total and average.

```python

marks = {"Math": 80, "English": 70, "Physics": 90}

total = 0
for subject in marks:
    total = total + marks[subject]

average = total / len(marks)

print("Marks:", marks)
print("Total marks:", total)
print("Average marks:", average)
```

Q-7 Highest Scoring Student

student = {
"Ali": 85,
"Sara":92,
"Ahmed":78,
"Zara": 95
}

Find Highest scoring student.

```python

student = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zara": 95
}

highest_name = None
highest_score = 0

for name, score in student.items():
    if score > highest_score:
        highest_score = score
        highest_name = name

print("Student marks:", student)
print("Highest scoring student:", highest_name, "with", highest_score, "marks")
```

Q-8 Merge Dictionaries

    dict = {"a": 1, "b":2}
    dict = {"C": 3, "d":4}

Merge and print result.

```python

dict1 = {"a": 1, "b": 2}
dict2 = {"C": 3, "d": 4}

merged = {}
for key in dict1:
    merged[key] = dict1[key]

for key in dict2:
    merged[key] = dict2[key]

print("Dictionary 1:", dict1)
print("Dictionary 2:", dict2)
print("Merged dictionary:", merged)
```
